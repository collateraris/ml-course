```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor 

```

```python
rng = np.random.RandomState(1)
X = np.sort(5 * rng.rand(80, 1), axis=0)
y = np.sin(X).ravel()
y[::5] += 3 * (0.5 - rng.rand(16)) 
```

```python
regr_1 = DecisionTreeRegressor(max_depth=2)
regr_2 = RandomForestRegressor(max_depth=10, n_estimators=1000, criterion='friedman_mse')
regr_1.fit(X, y)
regr_2.fit(X, y)
```

```python
X_test = np.arange(0.0, 5.0, 0.01)[:, np.newaxis]
y_1 = regr_1.predict(X_test)
y_2 = regr_2.predict(X_test)
plt.figure(figsize=(16, 10))
plt.scatter(X, y, s=50, color='black', label='data')
plt.plot(X_test, y_1, color='green', label='max_depth=2', linewidth=3)
plt.plot(X_test, y_2, color='red', label='Random Forest', linewidth=3)
plt.xlabel('data') 
plt.ylabel('target')
plt.title('Decision Tree Regression')
plt.legend() 
plt.show()
```
