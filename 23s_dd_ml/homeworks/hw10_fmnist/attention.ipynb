```python
import wget
import os
import numpy as np
import torch, torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

```python
import re

with open('sequences.fasta', 'r') as iofile:
    text = iofile.readlines()
    
text_gen = [x.replace('\n', '') for x in text]
pat = re.compile(r'organism=.*$')
text_gen = [pat.sub('', h) for h in text_gen]
text_gen = [x.replace('-', '') for x in text_gen]
for x in text_gen:
    print(x)
```

```python
unique_symbols = set()                    # множество для уникальных символов
for seq in text_gen:
    unique_symbols.update(seq)
unique_symbols
```
